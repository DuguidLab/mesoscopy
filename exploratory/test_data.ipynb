```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.random.normal(70, 200, size=(500, 142, 142))
y = np.random.normal(65, 50, size=(500, 142, 142))
```

```python
plt.imshow(x[1, :, :], vmin=-500, vmax=500)
plt.show()
plt.imshow(y[1, :, :], vmin=-500, vmax=500)
```

```python
plt.plot(np.mean(x, axis=(1, 2)))
plt.plot(np.mean(y, axis=(1, 2)))
```

```python
plt.hist(np.mean(x, axis=(1, 2)))
plt.hist(np.mean(y, axis=(1, 2)))
```

```python
plt.plot(np.std(x, axis=(1, 2)))
plt.plot(np.std(y, axis=(1, 2)))
```

```python
plt.hist(np.std(x, axis=(1, 2)))
plt.hist(np.std(y, axis=(1, 2)))
```

```python
plt.plot(x[:, 0, 0])
plt.plot(y[:, 0, 0])
```

## Immitate dual channel recording with random frame order

```python
final_size = 1000

rand_idx = sorted(np.random.choice(final_size, 500, replace=False))
```

```python
merged = np.insert(x, rand_idx - np.arange(len(rand_idx)), y, axis=0)
```

```python
plt.imshow(merged[6, :, :], vmin=-500, vmax=500)
```
