```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.random.normal(70, 200, size=(500, 512, 512))
y = np.random.normal(65, 50, size=(500, 512, 512))
```

```python
plt.imshow(x[1, :, :])
```

```python
plt.plot(np.mean(x, axis=(1, 2)))
```

```python
np.std(x)
```

```python
plt.hist(np.std(x, axis=(1, 2)))
```

```python
plt.plot(np.mean(x, axis=(1, 2)))
plt.plot(np.mean(y, axis=(1, 2)))
```

```python
plt.plot(x[:, 0, 0])
plt.plot(y[:, 0, 0])
```

```python
x = np.random.normal(70, 50, size=(500, 512, 512))
y = np.random.normal(65, 35, size=(500, 512, 512))

plt.plot(np.mean(x, axis=(1, 2)))
plt.plot(np.mean(y, axis=(1, 2)))
```

```python
plt.plot(x[:, 0, 0])
plt.plot(y[:, 0, 0])
```

```python
plt.imshow(x[1, :, :])
```

```python
plt.imshow(x.mean(axis=0))
```
